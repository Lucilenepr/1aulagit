```python
def depositar(saldo):
  deposito = float(input("digite o valor do deposito?"))
  print(f"saldo anterior,{saldo}")
  saldo = saldo + deposito
  print(f"saldo atual,{saldo}")
```

```python
def sacar(saldo):
  saque = float(input("digite o valor do saque?"))

  if saque < saldo:
    print(f"saldo anterior,{saldo}")
    saldo = saldo - saque
    print(f"saldo atual,{saldo}")
  else:
    print(f"saldo insulficiente: {saldo}")
  return saldo
```

```python

saldo = float(input("digite o saldo inicial?"))
print("1- depositar")
print("2- sacar")
print("3- saldo")
print("4- sair")
opcao = input("escolha uma opção: ")
if opcao == "1":
  depositar(saldo)
elif opcao == "2":
  sacar(saldo)
elif opcao == "3":
  print(saldo)
else:
  print("Encerrar sistema")


```
